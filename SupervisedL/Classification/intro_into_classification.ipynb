# Logistic Regression – Classification

## 🎯 1️⃣ Chno hiya Classification?

Classification hiya type dyal Machine Learning
fin target variable ma kaykonch continuous,
walakin categorical.

Exemples:
- Spam / Not Spam
- Yes / No
- Success / Failure
- Ychri / Ma ychrich

Output khaso ykoun:
    0 oula 1

---

## ❌ 2️⃣ 3lach ma nst3mlouch Linear Regression?

Linear Regression kat3tina:

    y = wx + b

Problem:
- t9der t3ti -2
- t9der t3ti 3.5
- t9der t3ti 100

Hada ma kaynfa3ch f classification,
7it bghina probability bin 0 w 1.

---

## ✅ 3️⃣ Fikrat Logistic Regression

Instead of predicting value,
kat predict:

    P(y = 1 | x)

Y3ni:
    Probability dyal class 1

W khas:

    0 ≤ p ≤ 1

---

## 🔁 4️⃣ Sigmoid Function

Bash n7awlo ay real number l probability,
kanst3mlo:

    σ(z) = 1 / (1 + e^(-z))

Fin:

    z = wx + b

Properties:
- ila z kbir bzaf → p ≈ 1
- ila z sghir bzaf → p ≈ 0
- ila z ≈ 0 → p ≈ 0.5

---

## ⚙️ 5️⃣ Kifach kaykhdem Model?

Step 1:
    z = wx + b

Step 2:
    p = sigmoid(z)

Step 3 (Decision rule):
    ila p ≥ 0.5 → class = 1
    ila p < 0.5 → class = 0

---

## 📊 6️⃣ Decision Boundary

Decision boundary hiya fin:

    p = 0.5

W hadchi kay3ni:

    wx + b = 0

F:
- 1D → point
- 2D → line
- ND → hyperplane

---

## 📉 7️⃣ Loss Function

F Regression kntst3mlo:
    MSE

Hna kanst3mlo:
    Log Loss (Binary Cross Entropy)

Formula:

    L = -[ y log(p) + (1 - y) log(1 - p) ]

3lach?
7it MSE ma mzyanach m3a probabilities.

---

## 🧠 8️⃣ Interpretation dyal Probability

Ila model 3tak:

    p = 0.83

Ma3naha:
    Kayn 83% chance ykoun class = 1

Ma kay3nish certainty.
Kay3ni confidence.

---

## 🎚️ 9️⃣ Threshold

Usually kanst3mlo:

    threshold = 0.5

3lach?
7it:
- p > 0.5 → class 1 probable aktar
- p < 0.5 → class 0 probable aktar

Walakin:
Threshold t9der tbdelha 7sab problem.

مثلاً:
- Medical diagnosis → n9dro ndiro 0.3
- Fraud detection → n9dro ndiro 0.7

---

## 🏁 🔟 Summary

- Logistic Regression = Classification algorithm
