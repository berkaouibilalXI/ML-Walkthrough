What Is a Machine Learning Model? (With Intuition + Plots)

```python
import numpy as np
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt
```

```python
model = LinearRegression()
```

```python
x = np.array([40, 60, 80, 100]).reshape(-1, 1) #feature (size)
y = np.array([80, 120, 160, 200]) #label (price k$)
```

```python
#test
model.fit(x,y)
```

```python
# Parameters
print("Slope (w):", model.coef_[0])
print("Intercept (b):", model.intercept_)
```

```python
#Predictions
X_line = np.linspace(40, 100, 100).reshape(-1, 1)
y_line = model.predict(X_line)
```

```python
# Plot
plt.scatter(x, y)
plt.plot(X_line, y_line)
plt.xlabel("House Size (m²)")
plt.ylabel("Price (k$)")
plt.title("Linear Regression: House Price Prediction")
plt.show()
```
