TESTING vs TRAINING

it is about how do we know if our model will work on new data?

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
import numpy as np
import matplotlib.pyplot as plt
```

```python
X = np.array([40, 50, 60, 70, 80, 90, 100]).reshape(-1, 1) #feature (size m²)
y = np.array([80, 100, 120, 140, 160, 180, 200]) #labels(prices in k$)
```

reshape is used to change the shape of an array without changing its data

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,y, test_size=0.3, random_state=42
)
```

Explanation (VERY IMPORTANT)

test_size=0.3
    30% of data → test set
    70% → training set
random_state=42
    Fixes randomness
    Ensures reproducibility (professional habit)

42 because it's the answer to life, the universe and everything haha, from the book "The Hitchhiker's Guide to the Galaxy"

```python
plt.scatter(X_train, y_train, color='red')
plt.scatter(X_test, y_test, color="green")
plt.xlabel("House sizes (m²)")
plt.ylabel("Price (k$)")
plt.title("Training vs Test Datas")
plt.show()
```

```python
model = LinearRegression()
model.fit(X_train, y_train)
```

The model never sees test data and this simulates real-world conditions

```python
X_line = np.linspace(40,100,100).reshape(-1, 1)
y_line = model.predict(X_line)
```

Smooth line for visualization
Predictions made using learned parameters

```python
plt.scatter(X_train, y_train, color='red')
plt.scatter(X_test, y_test, color='green')
plt.plot(X_line, y_line, color="black")
plt.xlabel("House Size (m²)")
plt.ylabel("Price (k$)")
plt.title("Linear Regression: Train vs Test")
plt.show()
```

Step 5: Evaluate on Test Data (First Metric)

```python
from sklearn.metrics import mean_squared_error
```

```python
y_pred = model.predict(X_test)
```

```python
mse = mean_squared_error(y_test, y_pred)
print("Mean Squared Error: ", mse)
```
